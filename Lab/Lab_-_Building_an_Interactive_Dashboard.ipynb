## 8. Create the analysis DataFrame

Select the features needed for the dashboard and use `.copy()` to create a DataFrame named `analysis`. Handle missing values in a way that is appropriate for the question. If values need clearer labels or corrected storage types, make those changes here.



```python
analysis = titanic[["Pclass", "Sex", "Fare", "Survived"]].dropna().copy()
```

**How did you handle missing values, and why was that choice appropriate?**

None of the four selected features (`Pclass`, `Sex`, `Fare`, `Survived`) contain any missing values in this dataset (confirmed by `.isnull().sum()` above), so calling `.dropna()` did not remove any rows here. ## 9. Create any calculated features

If the dashboard needs a calculated category, date component, percentage, rate, or other calculated feature, create it here. If no calculated feature is needed, leave the code cell blank and explain why.



```python
analysis["passenger_class"] = analysis["Pclass"].map({
    1: "1st class",
    2: "2nd class",
    3: "3rd class"
})

analysis["outcome"] = analysis["Survived"].map({
    0: "Did not survive",
    1: "Survived"
})
```

I created `passenger_class`, a readable label ("1st class", "2nd class", "3rd class") for the numeric `Pclass` codes, and `outcome`, a readable label ("Survived" / "Did not survive") for the binary `Survived` column. These labels let the dropdown options and any figure legends or hover text display understandable text instead of raw numeric codes.

## 13. Create the first static Plotly figure

Build the first figure before placing it in Dash. Give it a descriptive title and human-readable axis and legend labels.



```python
default_data = analysis[analysis["passenger_class"] == default_class]

survival_summary = (
    default_data.groupby("Sex")["Survived"].mean().reset_index()
)
survival_summary["survival_rate"] = survival_summary["Survived"] * 100

fig1 = px.bar(
    survival_summary,
    x="Sex",
    y="survival_rate",
    color="Sex",
    category_orders={"Sex": ["female", "male"]},
    color_discrete_map={"female": "mediumvioletred", "male": "steelblue"},
    title=f"Survival Rate by Sex — {default_class}",
    labels={"survival_rate": "Survival rate (%)", "Sex": "Sex"}
)
fig1.update_yaxes(range=[0, 100])
fig1.update_layout(showlegend=False)
fig1.show()
```

**What evidence should a viewer obtain from the first figure?**

The viewer should see, for the selected passenger class, what percentage of male passengers and what percentage of female passengers survived, making it easy to judge whether — and by how much — women had a higher survival rate than men within that class.


## 14. Create the second static Plotly figure

The second figure should add different but related evidence rather than repeat the first figure in another style.



```python
fig2 = px.box(
    default_data,
    x="Sex",
    y="Fare",
    color="Sex",
    category_orders={"Sex": ["female", "male"]},
    color_discrete_map={"female": "mediumvioletred", "male": "steelblue"},
    title=f"Fare Distribution by Sex — {default_class}",
    labels={"Fare": "Fare (£)", "Sex": "Sex"}
)
fig2.update_yaxes(range=[0, 520])
fig2.update_layout(showlegend=False)
fig2.show()
```

**What does the second figure add to the first figure's evidence?**

While the first figure shows differences in survival outcomes, the second figure shows differences in what male and female passengers actually paid for their tickets within the same class. This adds economic context: it lets the viewer see whether fare (a proxy for wealth or cabin location within a class) also differed by sex, alongside the survival-rate comparison.

## 16. Create reusable figure code

Write a function that accepts the selected control value, filters or summarizes `analysis`, creates both Plotly figures, and returns the figures in a consistent order. Fix axis ranges or category orders when they need to remain stable across selections.



```python
def make_figures(selected_class):
    subset = analysis[analysis["passenger_class"] == selected_class]

    survival_summary = subset.groupby("Sex")["Survived"].mean().reset_index()
    survival_summary["survival_rate"] = survival_summary["Survived"] * 100

    fig1 = px.bar(
        survival_summary,
        x="Sex",
        y="survival_rate",
        color="Sex",
        category_orders={"Sex": ["female", "male"]},
        color_discrete_map={"female": "mediumvioletred", "male": "steelblue"},
        title=f"Survival Rate by Sex — {selected_class}",
        labels={"survival_rate": "Survival rate (%)", "Sex": "Sex"}
    )
    fig1.update_yaxes(range=[0, 100])
    fig1.update_layout(showlegend=False)

    fig2 = px.box(
        subset,
        x="Sex",
        y="Fare",
        color="Sex",
        category_orders={"Sex": ["female", "male"]},
        color_discrete_map={"female": "mediumvioletred", "male": "steelblue"},
        title=f"Fare Distribution by Sex — {selected_class}",
        labels={"Fare": "Fare (£)", "Sex": "Sex"}
    )
    fig2.update_yaxes(range=[0, 520])
    fig2.update_layout(showlegend=False)

    return fig1, fig2
```

## 17. Build the layout

Create a Dash app and its layout. Include the title, purpose statement, interactive control, two empty `dcc.Graph` components with unique IDs, and short conclusion and limitation statements.



```python
app = Dash(__name__)

app.layout = html.Div([
    html.H1("Titanic Passenger Survival and Fare Explorer"),
    html.P(
        "This dashboard helps a maritime historian or museum exhibit "
        "designer compare survival rates and ticket fares between male "
        "and female passengers within each passenger class aboard the "
        "RMS Titanic (1912)."
    ),

    html.Label("Select a passenger class:"),
    dcc.Dropdown(
        id="class-dropdown",
        options=class_options,
        value=default_class,
        clearable=False
    ),

    dcc.Graph(id="survival-figure"),
    dcc.Graph(id="fare-figure"),

    html.H3("Conclusion"),
    html.P(
        "Within every passenger class, female passengers had a "
        "substantially higher survival rate than male passengers, and "
        "this gap was widest in 2nd and 3rd class."
    ),

    html.H3("Limitation"),
    html.P(
        "This dashboard describes only the 891 passengers included in "
        "this historical training dataset, a subset of everyone aboard "
        "the Titanic, so exact percentages should not be read as a "
        "complete record of all passengers and crew."
    )
])
```

## 18. Connect the callback

Connect the control's `value` to the `figure` property of both graphs. The callback function should call the reusable figure function and return its two figures in the same order as the two outputs.



```python
@app.callback(
    Output("survival-figure", "figure"),
    Output("fare-figure", "figure"),
    Input("class-dropdown", "value")
)
def update_figures(selected_class):
    return make_figures(selected_class)
```

## 19. Check the callback as a Python function

Call the callback or reusable figure function with at least two valid control values. Confirm that each call returns two Plotly figures without an error.



```python
for test_class in ["1st class", "2nd class", "3rd class"]:
    result_fig1, result_fig2 = update_figures(test_class)
    print(
        test_class,
        "->",
        type(result_fig1).__name__,
        type(result_fig2).__name__
    )
```

**After testing every control option, what did you revise or confirm about the dashboard's behavior?**

After testing all three dropdown options, I confirmed that both figures update correctly for every class, that the survival-rate y-axis stays fixed at 0-100% so the bar heights are directly comparable across classes, and that the fare y-axis stays fixed at 0-520 so the (much lower) 2nd and 3rd class fares are shown on the same comparable scale as 1st class fares, even though it means their boxes appear compressed near the bottom.


**Which design choice most improves the dashboard's readability or usefulness?**

Fixing both figures' y-axis ranges across every dropdown selection is the design choice that most improves usefulness. Without it, switching between classes would rescale each chart to that class's own range, which would make the bars and boxes look similarly sized across very different underlying values and could mislead a viewer into underestimating how much survival rate and fare actually differ between classes.
